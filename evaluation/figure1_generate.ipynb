```python
from pathlib import Path
import base64
import os
from PIL import Image
from io import BytesIO

def require_env_path(name: str) -> Path:
    """Return a required absolute path from an environment variable."""
    value = os.environ.get(name)
    if not value:
        raise RuntimeError(
            f"Set {name} to an absolute path before running this notebook."
        )
    path = Path(value).expanduser()
    if not path.is_absolute():
        raise RuntimeError(f"{name} must be an absolute path: {value!r}")
    return path.resolve()

PROJECT_ROOT = require_env_path("NPC_PROJECT_ROOT")
OUT = PROJECT_ROOT / "figures"
OUT.mkdir(parents=True, exist_ok=True)

PNG_OUT = OUT / "Figure1_RnO_PIXEL_EXACT.png"
PDF_OUT = OUT / "Figure1_RnO_PIXEL_EXACT.pdf"
SVG_OUT = OUT / "Figure1_RnO_PIXEL_EXACT.svg"

# Exact PNG bytes of the visually approved Figure 1.
PNG_B64 = """"""

png_bytes = base64.b64decode(PNG_B64)
PNG_OUT.write_bytes(png_bytes)

im = Image.open(BytesIO(png_bytes))
w, h = im.size

# Marginless PDF with the exact raster artwork.
im.convert("RGB").save(PDF_OUT, "PDF", resolution=96.0)

# SVG container embedding the exact raster artwork.
svg = f"""<?xml version="1.0" encoding="UTF-8" standalone="no"?>
<svg xmlns="http://www.w3.org/2000/svg"
     xmlns:xlink="http://www.w3.org/1999/xlink"
     width="{w}" height="{h}" viewBox="0 0 {w} {h}">
  <image x="0" y="0" width="{w}" height="{h}"
         preserveAspectRatio="none"
         href="data:image/png;base64,{PNG_B64}"/>
</svg>
"""
SVG_OUT.write_text(svg, encoding="utf-8")

print("Saved:")
print(PNG_OUT)
print(PDF_OUT)
print(SVG_OUT)
```
